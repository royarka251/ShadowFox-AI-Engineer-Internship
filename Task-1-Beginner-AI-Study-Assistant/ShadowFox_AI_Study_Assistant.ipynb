
# ShadowFox AI Engineer Internship
## Project: AI Study Assistant

An AI-powered student utility application built using the Gemini API and Gradio.

### Features
- Summarize Notes
- Generate Quiz Questions
- Explain Difficult Topics
- Input Validation and Error Handling


```python

import gradio as gr

def study_assistant(task, user_text):
    if not user_text or not user_text.strip():
        return "Please enter your notes or topic first."

    prompts = {
        "Summarize Notes": "Summarize these notes in simple bullet points:\n\n",
        "Generate Quiz": "Create 5 MCQs with four options, correct answers, and short explanations:\n\n",
        "Explain Topic": "Explain this topic in simple language with an example:\n\n"
    }

    try:
        response = client.interactions.create(
            model="gemini-3.8-flash",
            input=prompts[task] + user_text
        )
        return response.output_text or "No response received."

    except Exception as e:
        print("ACTUAL API ERROR:", type(e).__name__, str(e))
        return f"Request failed: {type(e).__name__}. Check the Colab output below."


with gr.Blocks(title="AI Study Assistant") as demo:
    gr.Markdown("# AI Study Assistant")
    gr.Markdown(
        "Summarize notes, generate quizzes, or explain topics with AI."
    )

    task = gr.Radio(
        ["Summarize Notes", "Generate Quiz", "Explain Topic"],
        value="Summarize Notes",
        label="Choose a Study Tool"
    )

    user_text = gr.Textbox(
        lines=8,
        label="Enter your Notes or Topic",
        placeholder="Paste your notes here..."
    )

    output = gr.Textbox(
        lines=10,
        label="AI Study Assistant Result"
    )

    with gr.Row():
        submit = gr.Button("Generate with AI", variant="primary")
        clear = gr.Button("Clear")

    submit.click(
        fn=study_assistant,
        inputs=[task, user_text],
        outputs=output
    )

    clear.click(
        fn=lambda: ("", ""),
        inputs=[],
        outputs=[user_text, output]
    )

print("Simple AI Study Assistant is ready!")
```
