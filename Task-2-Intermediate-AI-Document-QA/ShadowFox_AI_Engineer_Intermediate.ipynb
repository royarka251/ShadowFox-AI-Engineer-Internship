```python
from google.colab import files

print("Upload one or more PDF or TXT documents.")

uploaded = files.upload()

documents = {}

for filename, content in uploaded.items():
    if filename.lower().endswith((".pdf", ".txt")):
        documents[filename] = content
        print(f"Uploaded successfully: {filename}")
    else:
        print(f"Skipped unsupported file: {filename}")

print("\nTotal supported documents:", len(documents))
```

```python
from pypdf import PdfReader
from io import BytesIO

extracted_documents = {}

for filename, content in documents.items():
    try:
        if filename.lower().endswith(".pdf"):
            reader = PdfReader(BytesIO(content))
            text = "\n".join(
                page.extract_text() or ""
                for page in reader.pages
            )
        elif filename.lower().endswith(".txt"):
            text = content.decode("utf-8", errors="replace")
        else:
            continue

        text = text.strip()

        if text:
            extracted_documents[filename] = text
            print("File:", filename)
            print("Characters extracted:", len(text))
            print("\nExtracted text preview:")
            print(text[:1000])
        else:
            print("No readable text found:", filename)

    except Exception as e:
        print("Error reading", filename, ":", e)

print("\nTotal documents with readable text:",
      len(extracted_documents))
```

```python
def chunk_text(text, chunk_size=800, overlap=120):
    if chunk_size <= 0:
        raise ValueError("chunk_size must be positive")
    if overlap < 0 or overlap >= chunk_size:
        raise ValueError("overlap must be smaller than chunk_size")

    chunks = []
    start = 0

    while start < len(text):
        end = min(start + chunk_size, len(text))
        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        if end == len(text):
            break

        start = end - overlap

    return chunks


all_chunks = []
chunk_sources = []

for filename, text in extracted_documents.items():
    chunks = chunk_text(text)

    for chunk_id, chunk in enumerate(chunks):
        all_chunks.append(chunk)
        chunk_sources.append({
            "filename": filename,
            "chunk_id": chunk_id
        })

print("Total chunks:", len(all_chunks))

for i, chunk in enumerate(all_chunks):
    print(f"\n--- Chunk {i + 1} ---")
    print(chunk[:300])
```

```python
from sentence_transformers import SentenceTransformer

if not all_chunks:
    raise ValueError("No chunks found. Complete Step 5 first.")

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

chunk_embeddings = embedding_model.encode(
    all_chunks,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

print("Embedding shape:", chunk_embeddings.shape)
print("Embeddings generated successfully!")
```

```python
import faiss
import numpy as np

if len(all_chunks) == 0:
    raise ValueError("No chunks available.")

dimension = chunk_embeddings.shape[1]

vector_index = faiss.IndexFlatIP(dimension)
vector_index.add(
    np.asarray(chunk_embeddings, dtype=np.float32)
)

print("FAISS index created successfully!")
print("Indexed vectors:", vector_index.ntotal)
```

```python
def retrieve_context(question, top_k=3):
    question = question.strip()

    if not question:
        return []

    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype(np.float32)

    k = min(top_k, len(all_chunks))
    scores, indices = vector_index.search(query_embedding, k)

    results = []

    for score, idx in zip(scores[0], indices[0]):
        if idx < 0:
            continue

        results.append({
            "text": all_chunks[idx],
            "source": chunk_sources[idx]["filename"],
            "chunk_id": chunk_sources[idx]["chunk_id"],
            "score": float(score)
        })

    return results


question = "What are embeddings?"

results = retrieve_context(question)

for result in results:
    print("Source:", result["source"])
    print("Similarity:", round(result["score"], 3))
    print(result["text"])
    print("-" * 40)
```

```python
def build_grounded_prompt(question, results):
    if not question.strip():
        return "Please enter a question."

    if not results:
        return "No relevant document context was found."

    context = "\n\n".join(
        f"[Source: {r['source']}, chunk {r['chunk_id']}]\n{r['text']}"
        for r in results
    )

    return f"""Answer using only the context below.
If the answer is missing, say that it was not found
in the uploaded documents. Do not invent facts.

CONTEXT:
{context}

QUESTION:
{question}

ANSWER:"""


print(build_grounded_prompt(question, results))
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Answer generation model loaded successfully!")
print("Device:", device)
```

```python
def generate_answer(question, results):
    if not question.strip():
        return "Please enter a question."

    if not results:
        return "No relevant context found."

    context = "\n\n".join(
        item["text"] for item in results
    )

    prompt = f"""Answer using only the context.
If the answer is absent, say you could not find it.
Do not invent facts.

Context:
{context[:2500]}

Question: {question}
Answer:"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=120,
            do_sample=False
        )

    answer = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return answer.strip()


question = "What does an embedding represent?"
results = retrieve_context(question)

answer = generate_answer(question, results)

print("Question:", question)
print("Answer:", answer)
```

```python
def show_sources(results):
    if not results:
        print("No source passages retrieved.")
        return

    for i, result in enumerate(results, start=1):
        print(f"\n--- Source {i} ---")
        print("File:", result["source"])
        print("Chunk:", result["chunk_id"])
        print("Similarity:", round(result["score"], 3))
        print("Supporting text:")
        print(result["text"])


show_sources(results)
```

```python
import gradio as gr
import numpy as np
import faiss
from pypdf import PdfReader

def process_files_and_answer(files, question):
    if not files:
        return "Please upload a PDF or TXT file.", ""

    if not question or not question.strip():
        return "Please enter a question.", ""

    chunks = []
    metadata = []

    for path in files:
        try:
            if path.lower().endswith(".pdf"):
                reader = PdfReader(path)
                text = "\n".join(
                    page.extract_text() or ""
                    for page in reader.pages
                ).strip()

            elif path.lower().endswith(".txt"):
                with open(path, "r", encoding="utf-8",
                          errors="replace") as f:
                    text = f.read().strip()
            else:
                continue

            if not text:
                continue

            for i, chunk in enumerate(chunk_text(text)):
                chunks.append(chunk)
                metadata.append({
                    "source": path.split("/")[-1],
                    "chunk_id": i
                })

        except Exception as e:
            print(f"Could not process {path}: {e}")

    if not chunks:
        return "No readable text found in the uploaded files.", ""

    vectors = embedding_model.encode(
        chunks,
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype(np.float32)

    index = faiss.IndexFlatIP(vectors.shape[1])
    index.add(vectors)

    query_vector = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype(np.float32)

    k = min(3, len(chunks))
    scores, ids = index.search(query_vector, k)

    retrieved = []

    for score, idx in zip(scores[0], ids[0]):
        if idx >= 0:
            retrieved.append({
                "text": chunks[idx],
                "source": metadata[idx]["source"],
                "chunk_id": metadata[idx]["chunk_id"],
                "score": float(score)
            })

    answer = generate_answer(question, retrieved)

    sources = "\n\n".join(
        f"File: {r['source']}\n"
        f"Chunk: {r['chunk_id']}\n"
        f"Similarity: {r['score']:.3f}\n"
        f"{r['text']}"
        for r in retrieved
    )

    return answer, sources


app = gr.Interface(
    fn=process_files_and_answer,
    inputs=[
        gr.File(
            label="Upload PDF/TXT",
            file_count="multiple",
            file_types=[".pdf", ".txt"],
            type="filepath"
        ),
        gr.Textbox(
            label="Ask a question",
            placeholder="What does an embedding represent?"
        )
    ],
    outputs=[
        gr.Textbox(label="AI Answer", lines=4),
        gr.Textbox(label="Retrieved Source Context", lines=12)
    ],
    title="ShadowFox AI Engineer - Document Q&A Assistant",
    description="Upload documents and ask questions grounded in their text."
)

app.launch(share=True)
```
